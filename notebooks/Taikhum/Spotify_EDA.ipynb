```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

Reading all Files

```python
base_path = "/Volumes/project/datasets/spotify/"
```

```python
albums_df = spark.read.csv(
    base_path+"albums.csv",
    header=True,
    inferSchema=True
)

artists_df = spark.read.csv(
    base_path+"artists.csv",
    header=True,
    inferSchema=True
)

artwork_df = spark.read.csv(
    base_path+"artwork.csv",
    header=True,
    inferSchema=True
)

links_df = spark.read.csv(
    base_path+"links.csv",
    header=True,
    inferSchema=True
)

songs_df = spark.read.csv(
    base_path+"songs.csv",
    header=True,
    inferSchema=True
)
```

```python
charts_albums_df = spark.read.csv(
    base_path+"charts_albums_weekly.csv",
    header=True,
    inferSchema=True
)
```

```python
charts_artists_df = spark.read.parquet(
    base_path+"charts_artists_daily.parquet"
)

charts_songs_df = spark.read.parquet(
    base_path+"charts_songs_daily.parquet"
)
```

```python
datasets = {
    "Albums": albums_df,
    "Artists": artists_df,
    "Artwork": artwork_df,
    "Links": links_df,
    "Songs": songs_df,
    "Album Charts": charts_albums_df,
    "Artist Charts": charts_artists_df,
    "Song Charts": charts_songs_df
}


for name,df in datasets.items():
    print(
        name,
        "Rows:",
        df.count(),
        "Columns:",
        len(df.columns)
    )
```

```python
metadata = []

for name,df in datasets.items():

    metadata.append(
        (
            name,
            df.count(),
            len(df.columns)
        )
    )


metadata_df = spark.createDataFrame(
    metadata,
    ["Dataset","Rows","Columns"]
)

display(metadata_df)
```

```python
column_info=[]

for name,df in datasets.items():

    for c,t in df.dtypes:
        column_info.append(
            (
                name,
                c,
                t
            )
        )


column_df=spark.createDataFrame(
    column_info,
    [
     "Dataset",
     "Column",
     "Datatype"
    ]
)

display(column_df)
```

```python
for name,df in datasets.items():

    print("\n",name)

    numeric=[]
    categorical=[]

    for column,dtype in df.dtypes:

        if dtype in ['int','bigint','double','float']:
            numeric.append(column)

        else:
            categorical.append(column)


    print("Numeric:")
    print(numeric)

    print("Categorical:")
    print(categorical)
```

2] Null Analysis

```python
def null_count(df):

    return df.select(
        [
            count(
                when(
                    col(c).isNull(),
                    c
                )
            ).alias(c)

            for c in df.columns
        ]
    )


for name,df in datasets.items():

    print(name)

    display(
        null_count(df)
    )
```

3] Duplicate Analysis

```python
duplicate_summary=[]


for name,df in datasets.items():

    total = df.count()

    unique = df.dropDuplicates().count()

    duplicates = total - unique


    duplicate_summary.append(
        (
            name,
            total,
            duplicates
        )
    )


duplicate_df = spark.createDataFrame(
    duplicate_summary,
    [
        "Dataset",
        "Total Rows",
        "Duplicate Rows"
    ]
)

display(duplicate_df)
```

Top Songs by Total Streams

```python
top_songs = (
charts_songs_df
.groupBy(
    "track_name",
    "artist_names"
)
.agg(
    sum("streams").alias("total_streams")
)
.orderBy(
    desc("total_streams")
)
.limit(10)
)

display(top_songs)
```

Top Artists by Streams

```python
top_artists = (
charts_songs_df
.groupBy(
    "artist_names"
)
.agg(
    sum("streams").alias("total_streams")
)
.orderBy(
    desc("total_streams")
)
.limit(10)
)


display(top_artists)
```

Cleaning - Silver Layer

Remove Business Duplicates - 192 records

```python
charts_songs_df = (
    charts_songs_df
    .dropDuplicates(
        [
            "date",
            "country",
            "uri"
        ]
    )
)
charts_songs_df.count()
```

Songs Table Cleaning

```python
songs_df = (
    songs_df
    .filter(
        col("track_uri").isNotNull()
    )
)
```

```python
songs_df = songs_df.fillna(
    {
        "label":"Unknown",
        "artist_names":"Unknown"
    }
)
```

Albums Cleaning

```python
albums_df = albums_df.fillna(
    {
        "label":"Unknown"
    }
)
```

Artist Cleaning

```python
artists_df = (
    artists_df
    .filter(
        col("artist_uri").isNotNull()
    )
)
```

Charts Cleaning

```python
charts_songs_df = (
    charts_songs_df
    .filter(
        col("date").isNotNull() &
        col("country").isNotNull() &
        col("uri").isNotNull() &
        col("rank").isNotNull()
    )
)
```

```python
charts_artists_df = (
    charts_artists_df
    .filter(
        col("date").isNotNull() &
        col("uri").isNotNull()
    )
)
```

```python
charts_albums_df = (
    charts_albums_df
    .filter(
        col("date").isNotNull() &
        col("uri").isNotNull()
    )
)
```
