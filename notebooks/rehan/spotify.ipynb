```pyspark
artists = spark.read.csv(
    "s3://group-1-dbda/raw_data/artists.csv",
    header=True,
    inferSchema=True
)
```

```pyspark
artists.show(5)
```

```pyspark
artists.printSchema()
```

```pyspark
print("Rows:", artists.count())
```

```pyspark
print("Columns:", len(artists.columns))
```

```pyspark
print("Column Names:", artists.columns)
```

```pyspark
artists.describe().show()
```

```pyspark
from pyspark.sql.functions import col, when, count
```

```pyspark
artists.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in artists.columns
]).show()

print("Duplicate Rows:", artists.count() - artists.distinct().count())
```
