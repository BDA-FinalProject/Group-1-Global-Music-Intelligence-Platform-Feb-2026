```python
songs_schema = StructType([
    StructField("track_uri", StringType(), False),
    StructField("track_name", StringType(), True),
    StructField("artist_names", StringType(), True),
    StructField("artist_uris", StringType(), True),
    StructField("label", StringType(), True),
    StructField("release_date", DateType(), True),
    StructField("all_uris", StringType(), True)
])
```

**No exact duplicate records were identified in the charts_songs_df dataset. Each row represents a unique chart observation (based on the combination of date, country, song, and ranking information).
```python
# step5- Quartile Analysis
# Quartiles for Streams

quantiles = charts_songs_df.approxQuantile(
    "streams",
    [0.25, 0.50, 0.75],
    0.01
)

print("=" * 60)
print("Quartile Analysis - Streams")
print("=" * 60)
print("Q1 (25%)   :", quantiles[0])
print("Median     :", quantiles[1])
print("Q3 (75%)   :", quantiles[2])
```

1. Right-Skewed Distribution

Notice the large jump:

Q1      = 4,875
Median  = 12,453
Q3      = 44,771
Mean    = 67,800
Maximum = 30,987,370

The mean (67,800) is much higher than the median (12,453).

This indicates that a relatively small number of songs receive exceptionally high stream counts, pulling the average upward.

This is typical of music streaming platforms, where a few blockbuster tracks account for a large share of total streams.

2. Median vs Mean

Half of the chart entries receive fewer than approximately 12,453 streams.

Only a small fraction of songs generate hundreds of thousands or millions of streams.

3. Upper Quartile

The top 25% of chart entries receive more than 44,771 streams.

This threshold helps identify highly popular songs.

The Spotify streaming ecosystem follows a long-tail popularity distribution.

Most songs achieve moderate streaming volumes, while a relatively small number become viral or global hits, contributing disproportionately to overall platform streams.

This is expected behavior rather than a data quality issue.

```python
#using IQR method for outlier detection
Q1 = 4875.0
Q3 = 44771.0

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("=" * 60)
print("Streams Outlier Threshold")
print("=" * 60)
print("IQR          :", IQR)
print("Lower Bound  :", lower_bound)
print("Upper Bound  :", upper_bound)
```

```python
from pyspark.sql import functions as F

upper_bound = 104615

outlier_count = charts_songs_df.filter(
    F.col("streams") > upper_bound
).count()

total_rows = charts_songs_df.count()

outlier_percent = (outlier_count * 100.0) / total_rows

print("=" * 60)
print("Streams Outlier Analysis")
print("=" * 60)
print("Outlier Records :", outlier_count)
print("Outlier %       :", outlier_percent)
```

These records should not be considered erroneous data.

Spotify streaming data naturally follows a long-tail distribution, where a relatively small proportion of songs generate exceptionally high streaming volumes due to:

Viral popularity
Playlist recommendations
New album releases
Seasonal songs
Global chart-topping hits
Social media trends

Therefore, these observations represent high-performing songs rather than data quality issues.
