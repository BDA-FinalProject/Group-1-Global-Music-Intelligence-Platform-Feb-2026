```python
##########missing values check#######
from pyspark.sql.functions import col, when, count

def missing_value_report(df):

    return df.select([
        count(when(col(c).isNull(), c)).alias(c + "_NULL")
        for c in df.columns
    ])
```

```python
from pyspark.sql.functions import trim

def blank_report(df):

    return df.select([
        count(
            when(trim(col(c)) == "", c)
        ).alias(c + "_Blank")

        for c in df.columns
    ])
```

```python
# Year-wise chart records
from pyspark.sql import functions as F

yearly_records = (
    charts_songs_df
    .groupBy(F.year("date").alias("Year"))
    .agg(
        F.count("*").alias("Chart_Records")
    )
    .orderBy("Year")
)

yearly_records.show(20, truncate=False)
```

Insight 1
Spotify's chart dataset has expanded steadily from 2017 to 2024, indicating increasing platform activity and broader global coverage.
Insight 2
The volume of chart records stabilizes during 2024–2025, suggesting that the chart generation process has become consistent.
Insight 3
The lower record count in 2026 is due to incomplete yearly data rather than reduced streaming activity.

#################################
Year-wise Record Analysis: 
The number of chart records increased steadily from 3.44 million in 2017 to 5.14 million in 2024, reflecting growth in Spotify chart coverage and platform activity. The record count remained stable in 2025, indicating a mature and consistent charting process. The lower count observed in 2026 (1.97 million) is attributed to partial-year data availability and should not be interpreted as a decline in platform usage.

```python
# Year-wise total streams
from pyspark.sql import functions as F

yearly_streams = (
    charts_songs_df
    .groupBy(F.year("date").alias("Year"))
    .agg(
        F.sum("streams").alias("Total_Streams")
    )
    .orderBy("Year")
)

yearly_streams.show(truncate=False)
```

Answers the following question:
Which year generated the highest listening activity?"

Year-wise Streaming Trend:
Total annual streams increased consistently from 176.35 billion in 2017 to 455.10 billion in 2024, demonstrating sustained growth in music consumption on the platform. The highest streaming activity was observed in 2024, while 2025 showed a modest decline despite stable chart activity. The lower streaming volume recorded for 2026 reflects incomplete yearly data rather than reduced user engagement.

```python
# Month wise chart records
from pyspark.sql import functions as F

monthly_records = (
    charts_songs_df
    .groupBy(F.month("date").alias("Month"))
    .agg(
        F.count("*").alias("Chart_Records")
    )
    .orderBy("Month")
)

monthly_records.show(20, truncate=False)
```

Answers the question:
How many chart entries are generated in each calendar month (across all years)?

Across all years in the dataset, March contains the largest number of chart records.
This means:
more daily chart entries exist for March,
or more countries contributed records during March.
It does not necessarily mean users listened to more music in March.
To answer that, we need Month-wise Total Streams, which is our next analysis.

##
Month-wise Chart Record Distribution:
The number of chart records is distributed fairly evenly across calendar months, ranging from 3.35 million in June to 3.86 million in March. This indicates consistent monthly data coverage throughout the dataset and suggests that subsequent month-level analyses can be performed without significant concerns about data imbalance.

```python
#Month-wise total streams(aggregated across all the years)
from pyspark.sql import functions as F

monthly_streams = (
    charts_songs_df
    .groupBy(F.month("date").alias("Month"))
    .agg(
        F.sum("streams").alias("Total_Streams")
    )
    .orderBy("Month")
)

monthly_streams.show(truncate=False)
```

Answers the below question:
During which months do Spotify users consume the most music?

##
December appears to be the period with the highest music consumption.
Possible contributing factors include:
Holiday season
Christmas playlists
New Year celebrations
Increased leisure time
Family gatherings and travel
The dataset shows higher streaming activity in December, but it does not prove the exact cause. The holiday explanation is a reasonable business hypothesis.

##
March had the highest chart records.
March also has one of the highest stream totals.
This suggests March is consistently a strong month in this dataset.

Spotify users collectively streamed much less music in June than in December.
Again, the dataset shows the pattern; it does not establish the reason.

##
Now comparing the previous analysis.
Chart Records-

Noticeably different across months.
This tells us something important.
The number of chart records stays relatively stable.
The amount of listening changes considerably.

Monthly streaming activity exhibits noticeable seasonal variation. December recorded the highest total streams (265.66 billion), followed by March (260.76 billion), whereas June showed the lowest streaming activity (222.89 billion). Unlike chart record counts, which remain relatively stable throughout the year, total streams vary substantially, suggesting that user listening behavior changes across different months.
